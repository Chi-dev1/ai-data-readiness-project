```python
import pandas as pd
import numpy as np


from google.colab import files

uploaded = files.upload()

df = pd.read_csv(list(uploaded.keys())[0])

print("Original Data Info:")
print(df.info())

print("\nMissing Values:")
print(df.isnull().sum())

# -------------------------
# STEP 1: Simulate bad data
# -------------------------

df_dirty = df.copy()

# Introduce missing values
for col in df_dirty.columns[:2]:
    df_dirty.loc[df_dirty.sample(frac=0.2).index, col] = np.nan

# Duplicate rows
df_dirty = pd.concat([df_dirty, df_dirty.sample(frac=0.1)])

print("\nDirty Data Shape:", df_dirty.shape)

# -------------------------
# STEP 2: Show problem
# -------------------------

print("\nDirty Data Missing Values:")
print(df_dirty.isnull().sum())

# Example: numeric summary
print("\nDirty Data Summary:")
print(df_dirty.describe())

# -------------------------
# STEP 3: Clean data
# -------------------------

df_clean = df_dirty.drop_duplicates()

# Fill missing values
for col in df_clean.select_dtypes(include=np.number):
    df_clean[col].fillna(df_clean[col].mean(), inplace=True)

for col in df_clean.select_dtypes(include='object'):
    df_clean[col].fillna("Unknown", inplace=True)

print("\nClean Data Shape:", df_clean.shape)

# -------------------------
# STEP 4: Show improvement
# -------------------------

print("\nClean Data Missing Values:")
print(df_clean.isnull().sum())

print("\nClean Data Summary:")
print(df_clean.describe())

import matplotlib.pyplot as plt

df_dirty.select_dtypes(include=np.number).hist(figsize=(10,8))
plt.suptitle("Dirty Data Distribution")
plt.show()

df_clean.select_dtypes(include=np.number).hist(figsize=(10,8))
plt.suptitle("Clean Data Distribution")
plt.show()
```
